# Case Study Problem Statement

## RoutePulse Logistics Operations Intelligence System

RoutePulse is a logistics company that delivers customer shipments through distribution hubs located across different Indian cities.

The company stores its operational information across several related tables:

- customers,
- distribution hubs,
- drivers,
- service types,
- delivery slots,
- shipments,
- shipment payments, and
- monthly hub targets.

As the number of customers and shipments grows, the operations team is facing several challenges:

- customer cities must be compared with hub coverage,
- service types must be combined with available delivery slots,
- shipment fees must be compared with calculated averages,
- operational reports require information from several connected tables,
- hubs without assigned drivers must remain visible,
- actual delivery performance must be compared with operational targets,
- driver-supervisor relationships must be displayed,
- customer and shipment information needs cleaning and formatting,
- missing delivery and payment information must be handled safely,
- delivered shipments must be ranked,
- current shipments must be compared with earlier and later shipments, and
- cumulative and moving performance measures must be calculated without removing individual rows.

RoutePulse wants PostgreSQL queries that can transform its connected operational data into reliable reports.

As a junior SQL developer, your responsibility is to write the required queries for RoutePulse’s in-house operational reporting system.

The reporting system must be able to:

1. classify cities according to customer and hub coverage,
2. generate every service-type and delivery-slot combination,
3. identify delivered shipments with above-average delivery fees,
4. produce a complete connected shipment manifest,
5. identify hubs without assigned drivers,
6. compare actual delivered-shipment counts with assigned targets,
7. display drivers with their supervisors,
8. clean and transform customer, shipment, delivery, and payment information,
9. rank delivered shipments based on delivery fees,
10. compare consecutive shipments for each customer,
11. identify first and latest shipment values,
12. calculate customer-wise cumulative delivery fees, and
13. calculate moving averages while preserving row-level details.

| Concept Beginner-Friendly Meaning Where You Will Use It  |                                                                           |                                                                   |
| -------------------------------------------------------- | ------------------------------------------------------------------------- | ----------------------------------------------------------------- |
| Multi-Table Database                                     | Stores connected information in separate, focused tables.                 | Working with customers, hubs, drivers, shipments, and payments.   |
| Primary Key                                              | A column that uniquely identifies every row.                              | Identifying customers, hubs, drivers, and shipments.              |
| Foreign Key                                              | A column that connects a record with another table.                       | Connecting shipments with customers, drivers, hubs, and services. |
| Set Operation                                            | Combines or compares the outputs of multiple `SELECT` queries.            | Classifying customer and hub cities.                              |
| `UNION`                                                  | Combines results and removes duplicate rows.                              | Combining city-classification results.                            |
| `UNION ALL`                                              | Combines results and retains duplicate rows.                              | Combining independently classified city groups.                   |
| `INTERSECT`                                              | Returns rows present in both query results.                               | Finding cities containing customers and hubs.                     |
| `EXCEPT`                                                 | Returns rows from the first result that are absent from the second.       | Finding customer-only and hub-only cities.                        |
| Table Alias                                              | A temporary shorter name assigned to a table.                             | Writing readable multi-table queries.                             |
| Column Alias                                             | A temporary output name assigned using `AS`.                              | Producing the required result-column names.                       |
| Subquery                                                 | A query written inside another query.                                     | Comparing delivery fees with the calculated average.              |
| `CROSS JOIN`                                             | Combines every row from one table with every row from another.            | Generating service and delivery-slot combinations.                |
| `INNER JOIN`                                             | Returns rows with matching records in all joined tables.                  | Producing the connected shipment manifest.                        |
| `LEFT JOIN`                                              | Preserves every row from the left table.                                  | Including drivers without supervisors or hubs.                    |
| `RIGHT JOIN`                                             | Preserves every row from the right table.                                 | Including hubs without assigned drivers.                          |
| `FULL OUTER JOIN`                                        | Preserves matched and unmatched rows from both sides.                     | Comparing actual hub deliveries with assigned targets.            |
| Self-Join                                                | Joins a table with itself using separate aliases.                         | Connecting drivers with their supervisors.                        |
| `NULL`                                                   | Represents missing, unknown, or unavailable information.                  | Pending deliveries, missing payments, and unassigned records.     |
| `COALESCE`                                               | Returns the first non-`NULL` value.                                       | Replacing missing amounts, counts, and display labels.            |
| `CONCAT()`                                               | Combines multiple text values.                                            | Building customer names and shipment codes.                       |
| `SUBSTRING()`                                            | Extracts a selected part of text.                                         | Extracting usernames from customer email addresses.               |

| `LENGTH()`                                               | Returns the number of characters in text.                                 | Calculating email length.                                         |
| `ROUND()`                                                | Rounds a numeric value.                                                   | Formatting weights, averages, and percentages.                    |
| `CEIL()`                                                 | Rounds a value upward to the nearest integer.                             | Transforming shipment weights.                                    |
| `FLOOR()`                                                | Rounds a value downward to the nearest integer.                           | Transforming shipment weights.                                    |
| `ABS()`                                                  | Returns the non-negative size of a difference.                            | Calculating outstanding payment amounts.                          |
| Date Subtraction                                         | Calculates the number of days between two PostgreSQL dates.               | Calculating delivery duration.                                    |
| `EXTRACT()`                                              | Retrieves a selected part of a date.                                      | Extracting the booking year.                                      |
| `CASE`                                                   | Returns different outputs based on conditions.                            | Creating target and delivery classifications.                     |
| `NULLIF()`                                               | Returns `NULL` when two compared values are equal.                        | Preventing division-by-zero errors.                               |
| `CAST()`                                                 | Converts a value from one data type to another.                           | Creating text-based shipment codes.                               |
| Window Function                                          | Calculates across related rows without collapsing them.                   | Ranking shipments and analyzing customer shipment history.        |
| `OVER()`                                                 | Defines the rows used by a window function.                               | Every window-function query.                                      |
| `PARTITION BY`                                           | Divides rows into independent analytical groups.                          | Analyzing each customer’s shipments separately.                   |
| `ROW_NUMBER()`                                           | Assigns a unique sequential number to each row.                           | Numbering delivered shipments.                                    |
| `RANK()`                                                 | Gives tied values the same rank and leaves a rank gap.                    | Ranking shipments with equal fees.                                |
| `DENSE_RANK()`                                           | Gives tied values the same rank without leaving a gap.                    | Creating continuous fee ranks.                                    |
| `LAG()`                                                  | Retrieves a value from a previous row.                                    | Finding the previous shipment fee.                                |
| `LEAD()`                                                 | Retrieves a value from a following row.                                   | Finding the next shipment fee.                                    |
| `FIRST_VALUE()`                                          | Returns the first value in an ordered window.                             | Finding a customer’s first shipment fee.                          |
| `LAST_VALUE()`                                           | Returns the final value in the defined window frame.                      | Finding a customer’s latest shipment fee.                         |
| Running Total                                            | Calculates a cumulative value from the first row through the current row. | Calculating cumulative customer delivery fees.                    |
| Moving Average                                           | Calculates an average over the current row and nearby rows.               | Analyzing recent shipment-fee patterns.                           |
| Window Frame                                             | Defines the exact rows included in a window calculation.                  | Running totals, moving averages, and `LAST_VALUE()`.              |
